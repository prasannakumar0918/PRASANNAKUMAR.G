```python
import pandas as pd
import numpy as np
from IPython.display import display, HTML

class ColabDataPipeline:
    def __init__(self):
        self.df = None
        self.analysis_df = None

    def extract(self):
        """Step 1: Extract Data"""
        data = {
            'Emp_ID': [101, 102, 103, 104, 105, 101, 106, 107],
            'Name': ['Raj', 'Priya', 'Kumar', 'Anita', 'Bala', 'Raj', 'Divya', 'Karthik'],
            'Department': ['IT', 'HR', 'Finance', 'IT', 'Marketing', 'IT', 'Finance', 'HR'],
            'Projects_Completed': [5, 3, '8', 2, 12, 5, 4, np.nan],
            'Satisfaction_Score': [85, 90, 45, 120, 70, 85, 60, 55],
            'Attendance_Percentage': [95, 88, 40, 92, 60, 95, 75, np.nan]
        }
        self.df = pd.DataFrame(data)
        print("🔹 [STEP 1: EXTRACT] Raw Data Loaded Successfully:")
        display(self.df)
        return self  # Method chaining-kku 'self' return panrathu romba mukkiyam!

    def clean(self):
        """Step 2: Clean Data"""
        # Duplicates remove panrathu
        self.df = self.df.drop_duplicates()

        # Numeric conversion & missing values fill panrathu
        self.df['Projects_Completed'] = pd.to_numeric(self.df['Projects_Completed'])
        self.df['Projects_Completed'] = self.df['Projects_Completed'].fillna(self.df['Projects_Completed'].median())
        self.df['Attendance_Percentage'] = self.df['Attendance_Percentage'].fillna(self.df['Attendance_Percentage'].mean())

        # Invalid values correct panrathu
        self.df['Satisfaction_Score'] = self.df['Satisfaction_Score'].apply(lambda x: 100 if x > 100 else x)

        print("🔹 [STEP 2: CLEAN] Data Cleaned Successfully:")
        display(self.df)
        return self

    def transform(self):
        """Step 3: Transform Data"""
        bins = [0, 50, 75, 100]
        labels = ['Needs Improvement', 'Good', 'High Performer']
        self.df['Performance_Category'] = pd.cut(self.df['Satisfaction_Score'], bins=bins, labels=labels)

        print("🔹 [STEP 3: TRANSFORM] Categories Added Successfully:")
        display(self.df[['Name', 'Department', 'Satisfaction_Score', 'Performance_Category']])
        return self

    def analyze(self):
        """Step 4: Analyze Data"""
        self.analysis_df = self.df.groupby('Department')['Satisfaction_Score'].mean().reset_index()

        print("🔹 [STEP 4: ANALYZE] Department Insights Generated:")
        display(self.analysis_df)
        return self

    def load(self):
        """Step 5: Load / Save Data"""
        self.df.to_csv('cleaned_employee_data.csv', index=False)
        self.analysis_df.to_csv('department_analysis_report.csv', index=False)

        print("🔹 [STEP 5: LOAD] Pipeline Completed & Files Saved!")
        print("📁 Saved files: 'cleaned_employee_data.csv' & 'department_analysis_report.csv'")

# ==========================================
# PIPELINE EXECUTION (Method Chaining Method)
# ==========================================
print("🚀 Pipeline Execution Started...\n")
pipeline = ColabDataPipeline()

# Orey line-la ella steps-aiyum sequence-ah run panrathuthaan "Pipeline Method"!
pipeline.extract().clean().transform().analyze().load()
```
