```
#find the largest of three number
def largest_of_three(a,b,c):
    largest=a
    if b>largest:
        largest=b
    if c>largest:
        largest=c
    return largest
print(largest_of_three(12,45,30))
```

```
#check armstrong number
def is_armstrong(n):
  digits=str(n)
  power=len(digits)
  total=sum(int(digit)**power for digit in digits)
  return total==n
num=int(input('Enter a number:'))
print('Armstrong Number'if is_armstrong(num) else 'Not an Armstrong Number')
```

```
 #remove duplicates from a list
def remove_duplicates(lst):
  seen=set()
  result=[]
  for item in lst:
    if item not in seen:
      seen.add(item)
      result.append(item)
  return result
print(remove_duplicates([1,2,3,4,5]))
```

```
#find the secound largest number in a list
def second_largest(lst):
  unique=list(set(lst))
  unique.sort(reverse=True)
  return unique[1]
print(second_largest([10,25,3,47,8]))
```

```
#check if two strings are anagrams
def is_anagrams(s1,s2):
  s1=s1.replace('','').lower()
  s2=s2.replace('','').lower()
  return sorted(s1)==sorted(s2)
print(is_anagrams('listen','silent'))
print(is_anagrams('hello','world'))
```

```
#fizzbuzz
def fizzbuzz(n):
  result=[]
  for i in range(1,n+1):
    if i%15==0 and i%5==0:
      result.append('FizzBuzz')
    elif i%3==0:
      result.append('Fizz')
    elif i%5==0:
      result.append('Buzz')
    else:
      result.append(str(i))
  return result
print(''.join(fizzbuzz(15)))
```

```
#find the gcd (hcf)of two numbers
def gcd(a,b):
  while b:
    a,b=b,a%b
    return a
print(gcd(48,18))
import math
print(math.gcd(48,18))
```

```
#cound the frequency of each element in a list
def count_frequency(lst):
  freq={}
  for item in lst:
   freq[item]=freq.get(item,0)+1
   return freq
print(count_frequency(['a','b','a','c','b','a']))
from collections import Counter
print(Counter(['a','b','a','c','b','a']))
```
